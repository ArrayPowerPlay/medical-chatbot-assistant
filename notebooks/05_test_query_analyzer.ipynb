```python
import os
import sys
import json

# Add root project to sys.path so we can import src
project_root = os.path.abspath(os.path.join(os.getcwd(), '..'))
if project_root not in sys.path:
    sys.path.insert(0, project_root)

from src.query.query_analyzer import QueryAnalyzer

# Initialize QueryAnalyzer
analyzer = QueryAnalyzer()
print("QueryAnalyzer initialized successfully!")
```

```python
def test_and_validate(query: str, history=None):
    print(f"\n{'='*50}")
    print(f"TESTING QUERY: '{query}'")
    print(f"{'='*50}")
    
    # Run the analyzer
    result = analyzer.analyze(query, history=history)
    
    # Print JSON output
    print("\n--- LLM JSON OUTPUT ---")
    print(json.dumps(result, indent=2, ensure_ascii=False))
    
    # Validate format
    expected_keys = {"rewritten_query", "diseases", "symptoms", "drugs", "intents"}
    missing_keys = expected_keys - set(result.keys())
    
    print("\n--- FORMAT VALIDATION ---")
    if not missing_keys:
        print("✅ SUCCESS: Output contains all required keys.")
        print(f"✅ SUCCESS: Intents parsed properly: {result.get('intents')}")
    else:
        print(f"❌ ERROR: Missing keys: {missing_keys}")

# Test 1: Simple query
# test_and_validate("What are the symptoms and treatments for type 2 diabetes?")

# Test 2: Query with history context (pronoun resolution)
history = [
    {"role": "user", "content": "Tell me about Metformin"},
    {"role": "assistant", "content": "Metformin is a medication for type 2 diabetes."}
]
test_and_validate("What are its common side effects?", history=history)
```
